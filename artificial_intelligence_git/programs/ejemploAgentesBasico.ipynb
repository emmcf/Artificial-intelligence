##Un agente inteligente básico
PERCEPCIÓN → DECISIÓN → ACCIÓN

```python
import random
class SimpleAgent:
    def __init__(self,name):
        self.name=name
    def decide_move(self):
        return random.choice(["left","right"])
agente=SimpleAgent("Agente A")
print(f"{agente.name} decide el movimiento {agente.decide_move()}.")
agente2=SimpleAgent("Agente B")
print(f"{agente2.name} decide el movimiento {agente2.decide_move()}.")
agente3=SimpleAgent("Agente C")
print(f"{agente3.name} decide el movimiento {agente3.decide_move()}.")
```

##Agente reactivo

```python
class RobotAspiradora:
    def percibir(self, ambiente):
        return ambiente
    def decidir(self, ambiente):
        if ambiente == "sucio":
            return "aspirar"
        elif ambiente == "obstaculo":
            return "girar"
        elif ambiente == "limpio":
            return "avanzar"
        else:
            return "detenerse"
    def actuar(self, accion):
        acciones = {
            "aspirar": "Aspirando...",
            "girar": "Girando...",
            "avanzar": "Avanzando...",
            "detenerse": "Deteniéndose..."
        }
        print(acciones[accion])
robot = RobotAspiradora()
ambientes = ["limpio","sucio","limpio", "obstaculo"]
for ambiente in ambientes:
    print(f"\nEntorno: {ambiente}")
    percepcion = robot.percibir(ambiente)
    accion = robot.decidir(percepcion)
    robot.actuar(accion)
```

##Agente con objetivo
En este caso, el agente ya tienen un objetivo, en este ejemplo será llegar a un punto determinado de un lugar.

```python
class Robot:
    def __init__(self, posicion, objetivo):
        self.posicion = posicion
        self.objetivo = objetivo
    def decidir(self):
        x, y = self.posicion
        gx, gy = self.objetivo
        if x < gx:
            return "derecha"
        elif y < gy:
            return "abajo"
        else:
            return "detenerse"
    def actuar(self, accion):
        x, y = self.posicion
        if accion == "derecha":
            x += 1
        elif accion == "abajo":
            y += 1
        self.posicion = (x, y)
robot = Robot(
    posicion=(0, 0),
    objetivo=(4, 4)
)
while robot.posicion != robot.objetivo:
    accion = robot.decidir()
    print(f"Posición: {robot.posicion} → Acción: {accion}" )
    robot.actuar(accion)
print(f"Objetivo alcanzado: {robot.posicion}")
```

##MultiAgente colaborativo

```python
class Agente:
    def __init__(self, nombre, capacidad):
        self.nombre = nombre
        self.capacidad = capacidad
    def puede_realizar(self, tarea):
        return tarea in self.capacidad
    def realizar(self, tarea):
        print( f"{self.nombre} realiza '{tarea}'" )
robot1 = Agente( "Robot-A", ["transportar", "recoger"])
robot2 = Agente( "Robot-B", ["transportar", "inspeccionar"])
robot3 = Agente( "Robot-C", ["inspeccionar", "recoger"])
robot1.puede_realizar("transportar")
robot1.realizar("transportar")
robot2.puede_realizar("inspeccionar")
robot2.realizar("inspeccionar")
robot3.puede_realizar("recoger")
robot3.realizar("recoger")
```

```python
tarea = "inspeccionar"
for robot in [robot1, robot2, robot3]:
    if robot.puede_realizar(tarea):
        robot.realizar(tarea)
```

##Comunicacion entre agentes

```python
class Agente:
    def __init__(self, nombre):
        self.nombre = nombre
    def enviar_mensaje(self, receptor, mensaje):
        print(f"{self.nombre} → {receptor.nombre}: {mensaje}" )
    def recibir_mensaje(self, emisor, mensaje):
        print(f"{self.nombre} recibió de {emisor.nombre}: {mensaje}" )
robotA = Agente("Robot-A")
robotB = Agente("Robot-B")
robotA.enviar_mensaje( robotB, "Encontré un obstáculo en la zona norte.")
robotB.recibir_mensaje( robotA, "Encontré un obstáculo en la zona norte.")
```

##Coordinacion de agentes

```python
class Robot:
    def __init__(self, nombre, zona):
        self.nombre = nombre
        self.zona = zona
    def explorar(self):
        print( f"{self.nombre} explora la zona {self.zona}" )
    def informar(self, mensaje):
        print(f"{self.nombre}: {mensaje}" )
robots = [ Robot("Robot-A", "Norte"), Robot("Robot-B", "Centro"), Robot("Robot-C", "Sur")]
for robot in robots:
    robot.explorar()
robots[0].informar("Zona Norte despejada.")
robots[1].informar( "Se detectó un obstáculo.")
robots[2].informar( "Zona Sur despejada.")
```

##Sistemas ubicuos

```python
class AulaInteligente:
    def __init__(self,temperatura, personas,luz):
        self.temperatura = temperatura
        self.personas = personas
        self.luz = luz
    def decidir(self):
        if self.personas == 0:
            return "apagar_todo"
        if self.temperatura > 25:
            return "encender_aire"
        if self.luz < 40:
            return "encender_luz"
        return "mantener"
aula = AulaInteligente( temperatura=28, personas=25, luz=30)
accion = aula.decidir()
print("Acción:", accion)
```

##Otro ejemplo de un sistema ubicuo

```python
class Sensor:
    def __init__(self, nombre, valor):
        self.nombre = nombre
        self.valor = valor
    def leer(self):
        return self.valor
temperatura = Sensor("Temperatura",28)
personas = Sensor("Personas",20)
luz = Sensor("Iluminación",30)
print(temperatura.nombre,":",temperatura.leer())
print(personas.nombre, ":", personas.leer())
print(luz.nombre, ":",luz.leer())
```

```python
class AgenteAula:
    def decidir( self,temperatura, personas,luz):
        acciones = []
        if personas == 0:
            acciones.append( "Apagar dispositivos" )
        else:
            if temperatura > 25:
                acciones.append( "Encender aire acondicionado" )
            else:
              acciones.append( "Mantener aire acondicionado" )
            if luz < 40:
                acciones.append( "Encender iluminación" )
            else:
                acciones.append( "Mantener iluminación" )
        return acciones
```

```python
agente = AgenteAula()
acciones = agente.decidir( temperatura=28,  personas=20, luz=30)
for accion in acciones:
    print("aula", accion)
```
